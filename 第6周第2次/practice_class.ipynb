```python
class Student(object):
    def __init__ (self, name, score):
        self.name = name
        self.score = score
        print("========")

    def print_name_score(self):
        print("{%s}:{%d}".format(self.name,self.score))
```

```python
M = Student("Michal",98)
C = Student("Cathy",100)
J = Student("Jack",70)
```
