```python

def list_median(input_list):
	
	if len(input_list) == 0:
		return None
	else:
		input_list.sort()
		if len(input_list) % 2 == 1:
			median_index =  (len(input_list)+1)//2  -1
			return input_list[median_index]
		else:
			median_index = len(input_list) // 2 -1
			return (input_list[median_index] +input_list[median_index+1]) / 2

print(list_median([1,2,3,4,5]))
		
```

```python
def sum_int_list(input_list, target):
	result = []
	for i in input_list:
		for j in input_list[input_list.index(i):]:
			if i + j == target:
				result.append((input_list.index(i),input_list.index(j)))
	result.sort()
	return result


print(sum_int_list([1,2,3],5))
```

```python
def count_sum_int_float(input_list):
	num_list = []
	for i in input_list:
		if isinstance(i,float) or isinstance(i,int):
			num_list.append(i)
	if len(num_list) > 0:
		result = (0,None)
	else:
		result = (len(num_list),sum(num_list))
	return result
```

```python
def string_sorting(string_list):
	for i in string_list:
		temp = []
		temp_index = []
		if isinstance(i,str):
			temp.append(i)
			temp_index.append(string_list.index(i))
		else:
			continue
	temp.sort()
	for j in temp_index:
		string_list[j] = temp[temp_index.index(j)]
	return string_list
```
