```python

def list_median(input_list):
	
	if len(input_list) == 0:
		return None
	else:
		input_list.sort()
		if len(input_list) % 2 == 1:
			median_index =  (len(input_list)+1)//2  -1
			return input_list[median_index]
		else:
			median_index = len(input_list) // 2 -1
			return (input_list[median_index] +input_list[median_index+1]) / 2

print(list_median([1,2,3,4,5]))
		
```
