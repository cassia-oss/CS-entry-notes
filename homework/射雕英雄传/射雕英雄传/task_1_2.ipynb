```python
import os
import jieba
import csv
```

```python
folder = "D:\Vault\school\junior_last\python 与人工智能\python与人工智能-代码\homework\射雕英雄传\射雕英雄传"

jieba.load_userdict("D:\Vault\school\junior_last\python 与人工智能\python与人工智能-代码\homework\射雕英雄传\射雕英雄传词典.txt")

file_list = os.listdir(folder)
file_list.sort() 
words = []

for filename in file_list:
    if filename.endswith('.txt'):
        file_path = os.path.join(folder, filename)
        with open(file_path, "r", encoding="utf-8") as f:
            content = f.read()
        words += jieba.lcut(content)

words = list(filter(lambda x: len(x)>1,words))
```

```python
word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

word_count_list = list(word_count.items())

word_count_list = sorted(word_count_list,key = lambda x:-x[1])

for i in range(15):
    key , value = word_count_list[i]
    print("{0:<10}{1:>15}".format(key,value))

#word_count_list：list格式的元组集合，每个元组内部是一个词语和对应数值
#words是除去了标点符号的分词list（所有词）
#
```

```python
name_path = "D:\Vault\school\junior_last\python 与人工智能\python与人工智能-代码\homework\射雕英雄传\射雕英雄传人物.txt"
with open(name_path, "r", encoding="utf-8") as f:
    names = f.read()
names = list(names.split("\n"))
print(names)
```

```python

names_table = []
for filename in file_list:
    word_dic = {}
    file_path = os.path.join(folder, filename)
    with open(file_path, "r", encoding="utf-8") as f:
        content = f.read()
    for name in names:
        if not name in content:
            word_dic[name] = 0
        else:
            word_dic[name] = content.count(name)
    
    ls = list(word_dic.values())
    names_table.append(ls)

print(names_table)
```

```python
#转置矩阵，传入csv表格
data_reverse = [[names_table[i][j] for i in range(len(names_table))] for j in range(len(names_table[0]))]

print(data_reverse)

with open("names.csv",'w')as f:
    f_csv = csv.writer(f)
    f_csv.writerows(data_reverse)


```
