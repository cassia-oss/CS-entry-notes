```python
#最后一段：截取文字
def find_co_occurrence(name1, name2, k=0):
    if k >= len(positions[name1]):
        return f"{name1}不足{k+1} 次"

    start, end = positions[name1][k]

    # 截取原文上下文
    left = max(0, start - 80)
    right = min(len(txt), end + 80)
    context = txt[left:right]

    # 标记名字
    context = context.replace(name1, f"【{name1}】") 
    context = context.replace(name2, f"``{name2}``") 

    return context

print(find_co_occurrence("黄蓉","郭靖",100))
```
