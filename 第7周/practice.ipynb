```python
class Bus(object):
    def __init__(self,passengers):
        print("__init__ is called")
        self.passengers = passengers
    
    def pick(self,name):
        self.passengers.append(name)
    
    def drop(self,name):
        self.passengers.remove(name)
```

```python
class Train(Bus):
    def __init__(self,passengers):
        super().__init__(passengers)

    def pick(self,name):
        name += ' on train'
        self.passengers.append(name)
```

```python
Bus1 = Bus(['a','b','c'])
Bus1.pick('e')
Bus1.passengers
Bus1.drop('a')
Bus1.passengers
```

```python
Train1 = Train(['q','w','e'])
Train1.pick('r')
Train1.passengers
Bus.pick(Train1, 't')
Train1.passengers
```
