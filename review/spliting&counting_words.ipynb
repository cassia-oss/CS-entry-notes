# 读取文件
打开一般模式为：
```<variable> = open(<name>, <mode>)```
- ```<name>```：文件路径及名称
- ```<mode>```：打开模式；包括"r"（只读）"rt"（文本文件读取方式，属于默认文本模式）
- ```<variable>```：文件对象名


````python
#首先是打开文件部分：
## 打开 一个txt.文件
text_file = open("D:\Vault\school\junior_last\python 与人工智能\python与人工智能-代码\第3周第2次\高瓴人工智能学院简介.txt")
text = text_file.read()
text_file.close()
````

三个最基本的读取接口：read()、readline()、readlines()，

对应完全读取，读取一行，读取几行，

均为.read()方式使用


````python
# 读取
## .read()
f = open("123.txt", "rt", encoding="utf-8")
text = f.read()
print(text)
f.close()
## .readline()
f = open("123.txt", "rt", encoding="utf-8")
line1 = f.readline()
line2 = f.readline()
print(line1,end="")
print(line2,end="")
f.close()
## .readlines()
f = open("123.txt", "rt", encoding="utf-8")
lines = f.readlines()
lines = [line.strip() for line in lines]
print(lines)
f.close()
#或者采用with...as...的面向对象的方式创建：
with open("123.txt", "rt", encoding="utf-8") as f:
    text = f.read()
    print(text)
````

关闭：.close()

不然会占用磁盘内存

# 分词，并筛选段落中除去标点符号和“\n”的部分
一种很普遍的方法是使用jieba包，这个包可以自动切分段落为词语

````python
text_file = open("D:\Vault\school\junior_last\python 与人工智能\python与人工智能-代码\第3周第2次\高瓴人工智能学院简介.txt")
text = text_file.read()
text_file.close()
print(text)
````

````python
import jieba

words = jieba.lcut(text)

# 去掉单个干扰词语的方法：简单遍历
for word in words:
    if len(word) <= 1:
        words.remove(word)
    elif word == "\n":
        words.remove(word)
    else:
        continue
print(words)
````

````python
import jieba
words = jieba.lcut(text)

#使用列表推导式
words = [word.strip() for word in words if word.strip() and len(word.strip()) > 1] #列表推导式里使用if具有自带的删除功能
print(words)
````

````python
import jieba
words = jieba.lcut(text)

#使用filter
## 备注：filter的使用:filter(<判断函数>,<等待遍历的对象>)，将输出为"True"的对象保留，将输出为"False"的对象删去
words = list(filter(lambda x: len(x)>1,words))
print(words)
````
