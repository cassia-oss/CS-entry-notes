```python
import torch as t
import torch.nn as nn
```

```python
class MLPRRegression(nn.Module):
    def __init__(self,input_dim,hidden_dim):
        super().__init__()
        self.layer_1 = nn.Linear(input_dim,hidden_dim,bias = True)
        self.layer_2 = nn.Linear(hidden_dim,1)
        self.sig = nn.Sigmoid()

    def forward(self,x):
        x = self.layer_1(x)
        x = self.sig(x)
        x = self.layer_2(x)
        x = self.sig(x)
        return x

    def show_info(self):
        for name,parameter in self.named_parameters():
            print(name,"\n",parameter)
            print("------------------------------")
```

```python
a = MLPRRegression(3,4)
b = t.randn(3)

c = a(b)

a.show_info()
print(b)
print(c)
```

```python
class MLPModel():#这里原来是不用nn.Module的啊，因为前面已经把nn.Module处理过了
    def __init__(self,input_dim,hidden_dim,epoch,learning_rate):
        #初始化参数值
        self.epoches = epoch
        self.lr = learning_rate
        self.model = MLPRRegression(input_dim,hidden_dim)
        self.optimzer = t.optim.SGD(self.model.parameters(),lr = self.lr)
        self.loss_function = t.nn.MSELoss()

    def train(self,x,y):
        losses = []

        for i in range(self.epoches):
            prediction = self.model(x)
            loss = self.loss_function(prediction,y)
        
            self.optimzer.zero_grad()

            loss.backward()

            self.optimizer.step()

            losses.append(loss.item())#把数值从tensor中提取出来（并不是一个数一个数的训的，而是一起训的）

            if i % 500 == 0:
                print("epoch: {}, loss is: {}".format(i,loss.item()))

        return losses
    def test(self,x,y):
        prediction = self.model(x)
        testMSE = self.loss_function(prediction, y)

        return prediction,testMSE
```
