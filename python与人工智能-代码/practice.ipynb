```python
a = [1,2,[2,3]]
print(id(a))
print(id(a[2]))
print(id(a[2][0]))
a[2][1] = 1
print(id(a))
print(id(a[2]))
print(id(a[2][0]))
```

```python
l1 = [1,[2,3,4],(1,2,3)]
from copy import copy,deepcopy
l4 = l1.copy()
print(id(l1))
print(id(l4))
print(id(l1[1]))
print(id(l4[1]))
l4[1].append(1)
print("l1:",l1)
print("l4:",l4)
l4[-1]+=(1,)
print("l1:",l1)
print("l4:",l4)
print(id(l1))
print(id(l4))
l4[0] = 0
print("l1:",l1)
print("l4:",l4)
print(id(l1))
print(id(l4))
```
