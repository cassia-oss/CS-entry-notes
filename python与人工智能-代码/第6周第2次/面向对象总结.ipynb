```python
class Student(object):#前后两个下划线被成为魔法函数，为啥这样是可以的？
    
    def __init__(self, name, scores):
        self.name = name
        self.scores = scores
    
    def __iter__(self):
        return iter(self.scores)
    
    def __len__(self):
        return len(self.scores)
    
    def __getitem__(self, index):
        return self.scores[index]

    def __repr__(self):
        print("calling __repr__")
        return "[" + self.name + "\n" + ' '.join(map(str,scores)) + "]"
        
```

```python
scores = [80, 90, 100]
name = 'Tom'

tom = Student(name, scores)

print("Test __repr__") #直接使用print调用__repr__
print(tom)  #__repr__


print("Test __getitem__")
print(tom[1]) #__getitem__
print(tom.__getitem__(1))

print("Test __len__")
print(len(tom)) #__len__

print("Test __iter__")
for x in tom: #__iter__
    print(x)
```

```python
class A(object):
    def __init__(self):
        print("A is called.")

class B3(A):
    def __init__(self):
        super().__init__() #子类函数中需要调用父类的参数的时候，需要显式地调用父类的参数
        print("B3 is called.")


a = A()
b = B3()
```
