```python
import os
import jieba
import csv
```

```python
folder = "D:\Vault\school\junior_last\python 与人工智能\python与人工智能-代码\homework\射雕英雄传"

jieba.load_userdict(os.path.join(folder,"射雕英雄传词典.txt"))

file_list = os.listdir(os.path.join(folder,"射雕英雄传"))
file_list.sort() 
words = [] # 将全文按照词典进行分词，然后统一传入一个list

print(file_list)

for filename in file_list:
    if filename.endswith('.txt'):
        file_path = os.path.join(os.path.join(folder,"射雕英雄传"), filename)
        with open(file_path, "r", encoding="utf-8") as f:
            content = f.read()
        words += jieba.lcut(content)

words = list(filter(lambda x: len(x)>1,words))
```

```python
#计数，输出最大值
word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

word_count_list = sorted(word_count.items(),key = lambda x:-x[1])

for i in range(1):
    key , value = word_count_list[i]
    print("{0:<10}{1:>15}".format(key,value))

# word_count_list：list格式的元组集合，每个元组内部是一个词语和对应数值
# words是除去了标点符号的分词list（所有词）
```

```python
# 回答第二问，首先读取所有人物名称
name_path = os.path.join(folder,"射雕英雄传人物.txt")
with open(name_path, "r", encoding="utf-8") as f:
    names = f.read()
names = list(names.split("\n"))
print(names)
```

```python

names_table = []
for filename in file_list:
    name_dic = {}
    file_path = os.path.join(os.path.join(folder,"射雕英雄传"), filename)
    # 打开各个章节的文档
    with open(file_path, "r", encoding="utf-8") as f:
        content = f.read()
    for name in names:
        name_dic[name] = content.count(name)
    ls = list(name_dic.values())
    names_table.append(ls)
# 得到 行为章节，列为名字出现次数 的矩阵name_table
```

```python
#转置矩阵，传入csv表格
data_reverse = [[names_table[i][j] for i in range(len(names_table))] for j in range(len(names_table[0]))]

print(data_reverse)

with open("names.csv",'w')as f:
    f_csv = csv.writer(f)
    f_csv.writerows(data_reverse)
```
