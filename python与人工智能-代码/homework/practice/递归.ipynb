# 第 1 题：求列表元素和
1. 函数定义是求list中所有元素的和；
2. 边界条件是空列表，空列表返回0；
3. 递归关系可以定义为：n个元素的列表之和为第一个元素+剩下n-1个元素之和


```python
def list_sum(nums):
    if len(nums) == 0:
        return 0
    sum = nums[0] + list_sum(nums[1:])
    return sum

print(list_sum([1,2,3]))
```

# 第 2 题：求列表最大值
1. 函数定义是求list中最大值；
2. 边界条件是只有一个元素的时候返回该元素本身；
3. 递归关系可以定义为：n个元素的列表之和为第一个元素或者n-1个元素的最大值留取最大的一个。

```python
def list_max(nums):
    if len(nums) == 1:
        return nums[0]
    max_nums = max(nums[0],list_max(nums[1:]))
    return max_nums

print(list_max([12,3,4]))
```

# 第 3 题：判断回文串
1. 函数定义：判断一个string是否为回文的（前后对称）；
2. 边界条件是空字符串为True，单个字符串也为True；
3. 递归关系可以定义为：只判断首位两个元素是否为相同值，如果不相同，不用继续判断，如果相同，判断里边字符串是否为回问的。

```python
def is_palindrome(s):
    if len(s) < 2:
        return True
    is_pa = False
    if s[0] == s[-1]:
        is_pa = is_palindrome(s[1:-1])
    return is_pa

print(is_palindrome('12321'))
```

# 第 4 题：统计字符出现次数
1. 函数定义：统计ch在string中出现的次数；
2. 边界条件：|string| < |pattern|
3. 递归关系：如果前面m个字母为pattern，则+1 ；同时往后移动一个单位，使用递归。

```python
def count_char(s, ch):
    if len(s) < len(ch):
        return 0
    n = 0
    if s[:len(ch)] == ch:
        n = 1 + count_char(s[1:],ch)  #可重复版本，只往后移动一位；不可重复版本，往后移动len(ch)
    else:
        n = count_char(s[1:],ch)
    return n

print(count_char('ababa','aba'))
```

# 第 5 题：判断是否能被切分
1. 函数定义：判断字符串 string 是否能被切分成若干个字典中的词；
2. 边界条件：空字符‘’为0；
3. 递归关系：仍然是切分为前面的能够被切分，和后边的能够被切分。

```python
def can_cut(string, dictionary):
    if len(string) == 0:
        return True
    
    for i in range(1,len(string)+1):
        ex_str = string[:i]
        af_str = string[i:]
        if ex_str in dictionary:
            current_result = can_cut(af_str,dictionary)
            if current_result:
                return True

    return False

print(can_cut("abcabc", {"a": 1, "b": 1, "c": 1}) )
print(can_cut("abcabd", {"a": 1, "b": 1, "c": 1}) )
print(can_cut("南京市长江大桥", {"南京": 10, "南京市": 12, "长江": 10, "大桥": 10, "长江大桥": 25}))
```

# 第 6 题：数有几种切分方法
1. 函数定义：一个字符串能够有几种切分方法；
2. 边界条件：单位长度为0的字符串，有0种切分方法；
3. 递归关系：仍然是+1的方式切分字符串，如果一个前缀，后面需要跟相应的个数的后缀分类个数即可。

```python
def count_cut_ways(string, dictionary):
    if len(string) == 0:
        return 1
    
    ans = 0
    for i in range(1,len(string)+1):
        ex_str = string[:i]
        af_str = string[i:]
        if ex_str in dictionary:
            ans += count_cut_ways(af_str,dictionary)
        
    return ans

print(count_cut_ways("aaa", {"a":1, "aa":1}))
```

# 第 7 题：返回切分种类最多的方案
1. 函数定义：一个字符串最多能够按照字典切分为几个字符串；
2. 边界条件：没有切分种类的话就回归(None,0)，如果是长度为0的字符串，则回归([],0)；
3. 递归关系：同样是分为前后两个字符串，如果前面字符串在字典中，那么可以使用前面的字符串词数加后边字符串词数记录。

```python
def cut_more_words(string, dictionary):
    if len(string) == 0:
        return ([],0)
    
    best_cut = None
    best_len = 0
    for i in range(1,len(string)+1):
        ex_str = string[:i]
        af_str = string[i:]
        if ex_str in dictionary:
            rest_cut, rest_len = cut_more_words(af_str,dictionary)
            if rest_cut is None:
                continue
            current_cut = [ex_str] + rest_cut
            current_len = 1 + rest_len
            if current_len > best_len or best_cut is None:
                best_cut = current_cut
                best_len = current_len

    return (best_cut,best_len)


print(cut_more_words(
    "中华人民共和国",
    {"中华":1, "人民":1, "共和国":1, "中华人民":1, "中华人民共和国":1}
))
```

# 最终练习的题目：最多分词分数的大BOSS
实现word_cut(string, dictionary)函数。将字符串string分割为若干个dictionary中包含的词。

注意同一个字符串可能有不同的分词方式，我们需要选择其中最优的一种。因此，dictionary为一个字典，该字典的键为合法的词，对应的值为该词的得分（一个正整数）。一种分词方式的得分即为所有词的得分的和，最优的分词方式为得分和最大的分词方式。

word_cut函数需要返回一个包含最优分词结果(list)和最优分词结果的得分的元组。若输入为空字符串，应返回([], 0);若字符串不能被分割为字典中的词，应返回(None, 0)。

```python
def word_cut(string, dictionary):
    if len(string) == 0:
        return ([],0)

    best_cut = None
    best_score = 0
    for i in range(1,len(string)+1):
        ex_str = string[:i]
        af_str = string[i:]
        if ex_str in dictionary:
            rest_cut , rest_score = word_cut(af_str,dictionary)
            if rest_cut is None:
                continue
            current_cut = [ex_str] + rest_cut
            current_score = dictionary[ex_str] + rest_score
            
            if (
                best_cut is None
                or current_score > best_score
                or (
                    current_score == best_score
                    and len(current_cut) > len(best_cut)
                )
):
                best_cut = current_cut
                best_score = current_score

    return (best_cut,best_score)  


print(word_cut("abcabc", {"a": 1, "b": 1, "c": 1}))
print(word_cut("南京市长江大桥", {"南京": 10, "南京市": 12, "市长": 10, "长江": 10, "大桥": 10, "长江大桥": 25, "江大桥": 1})
)
print(word_cut("中华人民共和国", {"中华": 10, "人民": 10, "共和国": 15, "中华人民": 22, "中华人民共和国": 100}))
```

# 汉诺塔

```python
#汉诺塔移动过程
def hanoi(n, a, b, c):
    if n == 1:
        print( "move {} from {} to {}".format(n, a, c) )
    else:
        hanoi(n-1, a, c, b)
        print( "move {} from {} to {}".format(n, a, c) )
        hanoi(n-1, b, a, c)
hanoi(7, "A", "B", "C")

```
