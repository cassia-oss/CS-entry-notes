```python
import torch
import torch.nn as nn
import numpy as np
from PIL import Image
from torchvision import transforms
import torch.nn.functional as F
```

```python
image = Image.open('5.jpg').convert('RGB')
input = transforms.ToTensor()(image).unsqueeze(0)
print(input.shape)
#Conv-1
Conv1 = nn.Conv2d(in_channels=3, out_channels=6, kernel_size=5, stride=1, padding=0)
Nf1 = nn.ReLU()
output1 = Nf1(Conv1(input))
print(output1.shape)

for s,p in Conv1.named_parameters():
    print(s,p.shape)
```

```python
#Poo1-1，也相当于一个组件
Pool1 = nn.MaxPool2d((2,2), stride=2, padding=0)
output1 = Pool1(output1)
print(output1.shape)
```

```python
# A class implementation
class CNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(in_channels=3,out_channels=6,kernel_size=5,stride=1,padding=0)
        self.pool1 = nn.MaxPool2d((2,2),stride=2,padding=0)
        self.conv2 = nn.Conv2d(in_channels=6,out_channels=16,kernel_size=5,stride=1,padding=0)
        self.pool2 = nn.MaxPool2d((2,2),stride=2,padding=0)
        self.fc = nn.Linear(400,10)
        self.nf = nn.ReLU()

    def forward(self, x): #前向：完成一次从输入端到输出端的流动
        output1 = self.nf(self.conv1(x))
        print(output1.shape)
        output1 = self.pool1(output1)
        print(output1.shape)
        #####end of the first layer
        output2 = self.nf(self.conv2(output1))
        print(output2.shape)
        output2 = self.pool2(output2)
        print(output2.shape)
        #####end of the second layer

        # Flatten
        output3 = output2.view(x.shape[0],-1)#没有直接写1，因为X中可能有很多张图片
        print(output3.shape)
        # FC
        output3 = self.fc(output3)#过一次全连接层
        print(output3.shape)
        # SoftMax
        output = torch.softmax(output3,dim=1)
        return output

image = Image.open('5.jpg').convert('RGB')
input = transforms.ToTensor()(image).unsqueeze(0)
print(input.shape)

model = CNN()
output = model(input)
print(output)
```

```python
image2 = Image.open('3.jpg').convert('RGB')
input2 = transforms.ToTensor()(image2).unsqueeze(0)
print(input2.shape)

new_input = torch.cat([input,input2],0)
print(new_input.shape)

print("--------------------------")
new_output = model(new_input)
print(new_output)
print("--------------------------")
```
