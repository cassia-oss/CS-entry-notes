# 读取文件
打开一般模式为：
```<variable> = open(<name>, <mode>)```
- ```<name>```：文件路径及名称
- ```<mode>```：打开模式；包括"r"（只读）"rt"（文本文件读取方式，属于默认文本模式）
- ```<variable>```：文件对象名

三个最基本的读取接口：read()、readline()、readlines()，

对应完全读取，读取一行，读取几行，

均为.read()方式使用


````python
# 读取
## .read()
f = open("123.txt", "rt", encoding="utf-8")
text = f.read()
print(text)
f.close()
## .readline()
f = open("123.txt", "rt", encoding="utf-8")
line1 = f.readline()
line2 = f.readline()
print(line1,end="")
print(line2,end="")
f.close()
## .readlines()
f = open("123.txt", "rt", encoding="utf-8")
lines = f.readlines()
lines = [line.strip() for line in lines]
print(lines)
f.close()
#或者采用with...as...的面向对象的方式创建：
with open("123.txt", "rt", encoding="utf-8") as f:
    text = f.read()
    print(text)
````

关闭：.close()

不然会占用磁盘内存

# 分词，并筛选段落中除去标点符号和“\n”的部分
一种很普遍的方法是使用jieba包，这个包可以自动切分段落为词语

````python
text_file = open("D:\Vault\school\junior_last\python 与人工智能\python与人工智能-代码\第3周第2次\高瓴人工智能学院简介.txt")
text = text_file.read()
text_file.close()
print(text)
````

````python
import jieba

words = jieba.lcut(text)

# 去掉单个干扰词语的方法：简单遍历
for word in words:
    if len(word) <= 1:
        words.remove(word)
    elif word == "\n":
        words.remove(word)
    else:
        continue
print(words)
````

````python
import jieba
words = jieba.lcut(text)

#使用列表推导式
words = [word.strip() for word in words if word.strip() and len(word.strip()) > 1] #列表推导式里使用if具有自带的删除功能
print(words)
````

````python
import jieba
words = jieba.lcut(text)

#使用filter
## 备注：filter的使用:filter(<判断函数>,<等待遍历的对象>)，将输出为"True"的对象保留，将输出为"False"的对象删去
words = list(filter(lambda x: len(x)>1,words))
````

# 计数
接下来需要完成的计数工作，我们可以使用典型的if和for循环完成，并使用列表或字典两种形式，也可以使用jieba包中本身具有的独特性质和```disfault（<function>）```的特点来做。另外，还可以使用自带的Count函数。

````python
# 简单循环，使用list进行计数，但无法进行排序
word_list = []
count = []
for word in words:
    if word not in word_list:
        word_list.append(word)
        count.append(1)
    else:
        temp_index = word_list.index(word)
        count[temp_index] += 1

for i in range(15):
    print("{0:<5}{1:>10}".format(word_list[i],count[i]))
````

````python
# 仍然是使用循环计数的方法，不过这一次选择使用dic作为数据结构
word_count = {}
for word in words:
    word_count[word] = word_count.get(word,0) + 1

print(word_count)

#在这里可以对字典进行排序，使用sorted函数，按照value值进行排序
word_count_arrange = sorted(word_count,key = lambda x:-word_count[x]) #sorted函数返回的是按照key = 后规则排序后的列表
print(word_count_arrange)

for word in word_count_arrange[:15]:
    print("{0:<5}{1:>10}".format(word,word_count[word]))
````

````python
import jieba
from collections import defaultdict
positions = defaultdict(list)

for word,start_position,end_position in jieba.tokenize(text):
    if len(word)>1:
        positions[word].append((start_position,end_position))

print(positions)

word_count_arrange_defaultdic = sorted(positions.items(),key = lambda x:(-len(x[1]),x[0]))

for i in word_count_arrange_defaultdic[:15]:
    print("{0:<5}{1:>10}".format(i[0],len(i[1])))
````

````python
import jieba
from collections import Counter

word_count_Counter = Counter(words)

#使用Counter自带的排序函数
for word,_ in word_count_Counter.most_common(15):
    print("{0:<5}{1:>10}".format(word,word_count_Counter[word]))
````

在这一步主要进行的是计数和排序工作，而排序工作又很大程度上依赖于我是如何进行计数的，可以发现在这一过程中计数有很多种方法：

简单使用dic进行循环（可以添加get(,0)让其看上去更简洁）；

使用Counter自带的计数函数，这样就可以得到一个Counter(dic{})，后续可以直接使用.most_common(n)；

使用defauldict(list)，在没有键值的情况下生成一个defaultdic(<list>，{key_1:[]})可以对这个列表进行.append()操作，实现记录词语的起始位置这一目标。

对于这一排序函数，有两个有意思的地方：

输出是遍历对象按照排序方式重排后的list；

输入的可遍历对象如果是一个dic则之遍历key，如果使用.items()获取了其键值对的list则是在对形如[("A",nums),("B",nums)]的list做处理。
